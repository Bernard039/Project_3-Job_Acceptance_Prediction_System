### Step 1: Data Cleaning & Preprocessing

##### Filling Null Values

```python
# ssc_percentage

df['ssc_percentage']=df['ssc_percentage'].fillna(df['ssc_percentage'].median())
```

```python
# hsc_percentage 

df['hsc_percentage']=df['hsc_percentage'].fillna(df['hsc_percentage'].median())
```

```python
# career_switch_willingness

df['career_switch_willingness']=df['career_switch_willingness'].fillna(df['career_switch_willingness'].mode()[0])
```

```python
# relevant_experience

df['relevant_experience']=df['relevant_experience'].fillna(df['relevant_experience'].mode()[0])
```

```python
# job_role_match

df['job_role_match']=df['job_role_match'].fillna(df['job_role_match'].mode()[0])
```

```python
# notice_period_days

df['notice_period_days']=df['notice_period_days'].fillna(df['notice_period_days'].median())
```

```python
# layoff_history

df['layoff_history']=df['layoff_history'].fillna(df['layoff_history'].mode()[0])
```

```python
# employment_gap_months

df['employment_gap_months']=df['employment_gap_months'].fillna(df['employment_gap_months'].median())
```

```python
# Relocation Willingness

df['relocation_willingness']=df['relocation_willingness'].fillna(df['relocation_willingness'].mode()[0])
```

##### Cleaning Columns with Duplicate labels

```python
# Removing multiple Male & Female in Gender Column

df['gender'] = df['gender'].apply(lambda x: "Male" if x=="male"
                                  else "Female" if x=="female"
                                  else x
                                  )
```

```python
# Removing multiple Yes & No in "internship_experience" column

df['internship_experience']=df['internship_experience'].apply(lambda x:
                                                              "Yes" if x=="yes "
                                                              else "No" if x==" no"
                                                              else x
                                                              )
```

### Step 2: Feature Engineering

#### Derived Analytical Features 

```python
# Experience category (Fresher / Junior / Senior)

df['Experience category'] = df['years_of_experience'].apply(
    lambda x: "Fresher" if x == 0
    else "Junior" if 1 <= x <= 3
    else "Senior"
)
```

```python
# Qualification_percentage column

df['Qualification_percentage']=df[['ssc_percentage','hsc_percentage','degree_percentage']].median(axis=1)
```

```python
# Academic performance bands based on Qualification_percentage column

df['Academic_performance_bands']=df['Qualification_percentage'].apply(
    lambda x: "Excellent" if x >=75
    else "Poor" if x<40
    else "Good"
    )
```

```python
# Skills match level (Low / Medium / High)

df['Skills_match_level']=df['skills_match_percentage'].apply(
    lambda x: "High" if x>=90
    else "Low" if x<70
    else "Medium"
)
```

```python
# Interview Score based on Tech_Score, Aptitude_Score and Commun_Score

df['Interview_Score'] = df[['technical_score','aptitude_score','communication_score']].median(axis=1)
```

```python
# Interview performance category

df['Interview_performance_category']= df['Interview_Score'].apply(
    lambda x: "Excellent" if x >= 80
    else "Average" if x < 55
    else "Good"
)
```

####### Insights from Box Plot #######

Median comparison: The median score for Placed candidates is slightly higher (~70) than for Not Placed (~65).

Interpretation: Higher interview scores tend to align with placement, but the overlap shows that interview score alone doesn’t fully determine job acceptance.

```python
# Skills match percentage impact on placement

plt.Figure(figsize=(4,4))
sns.violinplot(data=df,x='status',y='skills_match_percentage')
plt.title("Skill Match vs Placement")
plt.xlabel('Placement')
plt.ylabel('Skill Percentage')
plt.show()
```

####### Insights from Violin Plot #######

Placed candidates: Their distribution is shifted slightly higher, with a median closer to the upper 70s. This suggests stronger skill alignment among those who got placed.

Not Placed candidates: Their distribution is more spread out, with a lower median (closer to mid-60s). This indicates more variability and generally lower skill match.

Overlap: Both groups share ranges between ~60–80, meaning skill percentage alone doesn’t fully determine placement. Other factors also matter.

Density: The “bulges” in the violin show where most candidates cluster. For example, if the Placed violin is thicker around 75–85, that’s where many successful candidates scored.

Relative comparison: Even though Tier 1 has the highest placements, its proportion of placed vs. not placed is similar to Tier 2. Tier 3 shows fewer opportunities overall.

```python
# Competition level vs job acceptance

cat2=pd.crosstab(df['competition_level'],df['status'])

plt.Figure(figsize=(4,4))
x=cat2.plot(kind='line',stacked=False)
plt.title("Competition Level vs Job Acceptance")
plt.show()
```

####### Insights from Line Plot #######

Low competition level: Both Placed and Not Placed counts are highest here. This suggests more candidates overall, with more opportunities but also more rejections.

High competition level: Fewer candidates overall, and placements are relatively low.

Medium competition level: Both lines dip compared to Low, showing fewer candidates and placements.

Trend: Placement numbers follow the same pattern as non-placement numbers — peaking at Low competition and dropping at Medium.

#### Scores

##### Accuracy Score

```python
from sklearn.metrics import accuracy_score

def evaluate_accuracy(models, x_test_scaler, y_test):
    results = {}
    for name, model in models.items():
        y_pred = model.predict(x_test_scaler)
        acc = accuracy_score(y_test, y_pred)
        results[name] = acc
    return results

# Trained models
models = {
    "Logistic Regression": Log_model,
    "Decision Tree": dt_model,
    "Naive Bayes": gnb_model,
    "SVM": svm_model,
    "KNN": knn_model,
}

accuracy_results = evaluate_accuracy(models, x_test_scaler, y_test)

for model, acc in accuracy_results.items():
    print(f"{model}: {acc:.4f}")
```

##### Evaluation Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_regressors(models, x_test_scaler, y_test):
    results = {}
    for name, model in models.items():
        y_pred = model.predict(x_test_scaler)
        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_test, y_pred)
        results[name] = {"MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2}
    return results

# Example: dictionary of trained regressor models
models = {
    "Random Forest": rf_model,
    "XGBoost": xgb_model,
    "LightGBM": lgbm_model,
    "CatBoost": catboost_model
}

results = evaluate_regressors(models, x_test_scaler, y_test)

# Print neatly
for model, metrics in results.items():
    print(f"{model}: MAE={metrics['MAE']:.4f}, MSE={metrics['MSE']:.4f}, RMSE={metrics['RMSE']:.4f}, R2={metrics['R2']:.4f}")
```
